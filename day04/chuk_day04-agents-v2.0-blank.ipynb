# ToolCallingAgent

`ToolCallingAgent` is an agent who rely on the LLM natvie ability to call functions/API. The LLM is give each tool's JSON schema, eg. name, description, parameter type, result type, etc. The LLM uses this information to call the functions.

## Differences between `CodeAgent` and `ToolCallingAgent`
| Aspect | CodeAgent | ToolCallingAgent |
|---|---|---|
| **Action format** | Python code snippet, written between `<code>` and `</code>` tags | Structured JSON tool calls through the provider's native function-calling API |
| **How tools appear to the LLM** | As Python functions it can call in code | As JSON schemas (name, description, parameter types) |
| **Tool calls per step** | Any number, combined with loops, conditionals and nested calls | One or more; each result goes back to the LLM before it can be used |
| **Parallel tool execution** | Sequential, as the code runs | Concurrent on a thread pool (`max_tool_threads`) |
| **State between steps** | Python variables persist across steps | Only what is in the message history |
| **Computation** | Built in (math, parsing, data transformation) | Only through tools; `add_base_tools=True` adds a `PythonInterpreterTool` |
| **Steps / LLM calls for multi-step tasks** | Usually fewer | Usually more round trips |
| **Executes model-written code?** | Yes | No |
| **Argument validation** | At runtime, when the code calls the tool | Against the tool schema, before the call |
| **Execution environment** | `LocalPythonExecutor` (restricted AST interpreter) or a remote sandbox | Your Python process calls the tools directly |
| **Security controls** | Import allow-list, operation limits, `timeout_seconds`, sandboxes (`docker`, `e2b`, `modal`, `blaxel`) | Limited to what your tools can do |
| **Specific parameters** | `additional_authorized_imports`, `executor_type`, `executor_kwargs`, `max_print_outputs_length`, `use_structured_outputs_internally`, `code_block_tags` | `max_tool_threads` |
| **Ending the run** | Calls `final_answer(...)` from code | Calls the `final_answer` tool |
| **Typical errors** | Syntax errors, runtime exceptions, forbidden imports | Wrong tool chosen, invalid arguments |
| **Model requirements** | A strong coding model | A model with reliable native function calling |
| **Works well with small local models?** | Often not (malformed code blocks) | Generally better |
| **Managed agents with a remote executor** | Not supported in 1.26 | N/A (no code execution) |
| **Best for** | Problem solving, chaining tools, data wrangling, reasoning over intermediate results | Dispatchers, simple independent API calls, strict validation, environments where code execution isn't allowed |
| **Shared by both** | Both inherit `MultiStepAgent`: ReAct loop, `max_steps`, memory, planning, step callbacks, `final_answer_checks`, streaming, managed agents, `instructions`, save/share | |

# Memory Retention
`reset` parameter is use to control the agent's memory retention between consecutive runs. Every time an agent is executed `run()`, it will forget all past conversation. It handles the `run()` as if it is the first request. This this the default behaviour `reset=True`.

By setting `reset=False`, we can ask the agent to use past conversation as context for the current conversation. This is essentially a follow-up quesiton paradigm.
